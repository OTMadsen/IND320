```python
#Next step is to retrieve the date from ENTSO and store it in Cassandra. Here I also define the time ranges for the plots, as well as defines the data that is going both in and out of Norway. 
import os
import time
import pandas as pd
from entsoe import EntsoePandasClient

client = EntsoePandasClient(
    api_key=os.environ["ENTSOE_API_KEY"]
)

start = pd.Timestamp("2024-11-01", tz="Europe/Oslo")
end = pd.Timestamp("2026-11-01", tz="Europe/Oslo")

connections = {
    "DE_LU": "NO_2 - DE_LU",
    "DK_1": "NO_2 - DK_1",
    "NL": "NO_2 - NL"
}

all_flows = []
failed_requests = []

for destination, connection_name in connections.items():
    for month_start in pd.date_range(
        start, end, freq="MS", inclusive="left"
    ):
        month_end = min(
            month_start + pd.DateOffset(months=1),
            end
        )

        for direction in ["out", "in"]:
            country_from = "NO_2" if direction == "out" else destination
            country_to = destination if direction == "out" else "NO_2"

            print(
                f"Downloading {connection_name}, "
                f"{month_start:%Y-%m}, {direction}"
            )

            try:
                series = client.query_crossborder_flows(
                    country_code_from=country_from,
                    country_code_to=country_to,
                    start=month_start,
                    end=month_end
                )

                if series is None or series.empty:
                    failed_requests.append(
                        (connection_name, str(month_start), direction)
                    )
                    continue

                batch = series.rename("value").reset_index()
                batch.columns = ["timestamp", "value"]

                batch["connection"] = connection_name
                batch["direction"] = direction

                all_flows.append(batch)

            except Exception as e:
                print(f"Request failed: {e}")
                failed_requests.append(
                    (connection_name, str(month_start), direction)
                )

            time.sleep(1)

if all_flows:
    electricity_df = pd.concat(all_flows, ignore_index=True)

    electricity_df["timestamp"] = pd.to_datetime(
        electricity_df["timestamp"], utc=True
    ).dt.tz_localize(None)

    electricity_df["value"] = pd.to_numeric(
        electricity_df["value"], errors="coerce"
    )

    electricity_df = electricity_df.dropna(
        subset=["timestamp", "value"]
    )

    electricity_df = electricity_df.drop_duplicates(
        subset=["timestamp", "connection", "direction"],
        keep="last"
    )

    print("\nTotal records:", len(electricity_df))
    print("First timestamp:", electricity_df["timestamp"].min())
    print("Last timestamp:", electricity_df["timestamp"].max())
    print("Connections:", electricity_df["connection"].unique())
else:
    print("No data downloaded.")

print("\nFailed or empty requests:", len(failed_requests))
print(failed_requests[:10])

```

```python

# Here we validate the downloaded ENTSO-E data

assert "electricity_df" in globals(), "Downloaded data is not in memory."
assert not failed_requests, f"Failed requests: {failed_requests}"

# Convert timestamps to Norwegian local time
local_months = (
    pd.to_datetime(electricity_df["timestamp"], utc=True)
    .dt.tz_convert("Europe/Oslo")
    .dt.to_period("M")
    .astype(str)
)

# Define the expected 24 months
expected_months = set(
    pd.period_range("2024-11", "2026-10", freq="M").astype(str)
)

actual_months = set(local_months)

assert expected_months <= actual_months, (
    f"Missing months: {sorted(expected_months - actual_months)}"
)

# Check all connections, directions and months
coverage = (
    electricity_df.assign(month=local_months)
    .groupby(["connection", "direction", "month"])
    .size()
    .reset_index(name="records")
)

assert len(coverage) == 144, (
    f"Expected 144 groups, found {len(coverage)}"
)

# Check for duplicate records
assert not electricity_df.duplicated(
    subset=["timestamp", "connection", "direction"]
).any(), "Duplicate records found."

# Display results
print("Total records:", len(electricity_df))
print("Coverage groups:", len(coverage))
print("All 24 months are represented for all connections and directions.")
```

```python
#Here we plot a pie chart for one of the connections in and out of Norway (In this case Germany)

import matplotlib.pyplot as plt
from pyspark.sql import functions as F 

el_energy_df = (
    spark.read
    .format("org.apache.spark.sql.cassandra")
    .options(keyspace="ind320_project", table="physical_flows")
    .load()
    .withColumn("year", F.year("timestamp"))  
)

pie_df_chart = (
    el_energy_df
    .filter(
        (F.col("year") == 2025) &
        (F.col("connection") == "NO_2 - DE_LU")
    )
    .groupBy("direction")
    .agg((F.sum(F.col("value").cast("double"))*0.25).alias("total")) #Now checks intervals every hour instead of every 15 minutes
    .toPandas()
)

if not pie_df_chart.empty:
    plt.pie(
        pie_df_chart["total"],
        labels=pie_df_chart["direction"],
        autopct="%1.1f%%"

    )

    plt.title("Transfers of Electricity between NO_2 (Norway) and DE_LU (Germany) in 2025")
    plt.show()
else:
    print("No data found. Please try again")
```

```python
#And now for the lineplot, also between Norway and Germany. 

line_df_chart = (
    el_energy_df
    .filter(
        (F.col("year") == 2025) &
        (F.month("timestamp") == 1) &
        (F.regexp_replace(F.col("connection"), " ", "") == "NO_2-DE_LU")
    )
    .orderBy("timestamp")
    .toPandas()
)

if not line_df_chart.empty:
    plt.figure(figsize=(12, 6))

    for direction, group in line_df_chart.groupby("direction"):
        plt.plot(
            group["timestamp"],
            group["value"],
            label=direction
        )

    plt.xlabel("Date")
    plt.ylabel("Electricity transfer (MW)")
    plt.title("Electricity transfers between NO_2 (Norway) and DE_LU (Germany) - January 2025")
    plt.legend()
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

else:
    print("No data found for Norway - Germany in January 2025.")
```

```python
"""
Now that this data has been processed and plotted, the final step is to move the Spark-Extracted data to MongoDB. 

Using some of the libraries from earlier, we are now going to move it. 

"""

import os 
from dotenv import load_dotenv
from pymongo import MongoClient

load_dotenv(".env")

mongo_uri = os.getenv("MONGODB_URI")

client_mongo = MongoClient(mongo_uri)
client_mongo.admin.command("ping")

print("Connected to Mongo!")

```

```python
#Now we just use the new updated MongoDB library to move the data from Spark to MongoDB. We read them in 5000 at a time, and then upload them to MongoDB.

from pymongo import ASCENDING
import pandas as pd

mongo_df = electricity_df[
    ["timestamp", "connection", "direction", "value"]
].copy()

mongo_df["timestamp"] = (
    pd.to_datetime(mongo_df["timestamp"], utc=True)
    .dt.tz_localize(None)
    )

assert not mongo_df.duplicated(
    ["timestamp", "connection", "direction"]
).any(), "Duplicate records found in Mongo DataFrame."

mongo_db = client_mongo["ind320_project"]

staging = mongo_db["electricity_flows_staging"]

staging.drop()

staging = mongo_db["electricity_flows_staging"]

staging.create_index(

    [

        ("connection", ASCENDING),
        ("timestamp", ASCENDING),
        ("direction", ASCENDING)
    ],
    unique=True
)

batch_size = 5000

for i in range(0, len(mongo_df), batch_size):
    batch = mongo_df.iloc[i:i + batch_size].to_dict("records")
    staging.insert_many(batch)


    print(
        f"Uploaded {min(i + batch_size, len(mongo_df))} of {len(mongo_df)}"
        f" / {len(mongo_df)} records"
    )

print("Staging upload success!")
```

```python
#Lastly, just to be sure everything is alright, we verify the data in Mongo

staging.rename(
    "electricity_flows",
    dropTarget=True
)

collection = mongo_db["electricity_flows"]

print(
    "Live MongoDB documents:",
    collection.count_documents({})
)

print("Updated two-year dataset is now working")
```
